# Chapter 23. Working with Time Series

## 23.1 Dates and Times in Python

### 23.1.1 Native Python: datetime module

```python
from datetime import datetime

# creation date
date = datetime(year=2021, month=7, day=4)
print(date)
```

```python
# 요일 출력
print(date.strftime('%A'))
```

### 23.1.2 dateutil module (string passing)

```python
from dateutil import parser

date = parser.parse("4th of July, 2021")
print(date)
```

### 23.1.3 NumPy의 datetime64

```python
import numpy as np

date = np.array('2021-07-04', dtype=np.datetime64)
print(date)
```

```python
# 벡터화 연산
print(date + np.arange(12))
```

### 23.1.4 4. Pandas의 Timestamp (최고의 선택!)

```python
import pandas as pd

date = pd.to_datetime("4th of July, 2021")
print(date)
print(date.strftime('%A'))

# NumPy 스타일 벡터화 연산
print(date + pd.to_timedelta(np.arange(12), 'D'))
```

## 23.2 Pandas Time Series: Indexing by Time

### Creation Datetimeindex

```python
# Timestamp로 indexing된 Series 생성
index = pd.DatetimeIndex(['2020-07-04', '2020-08-04',
                          '2021-07-04', '2021-08-04'])
data = pd.Series(range(4), index=index)
print(data)
```

### 날자 범위로 Slicing

```python
print(data['2020-07-04':'2021-07-04'])
```

### 년도만으로 Slicing

```python
print(data['2021'])
```

### pd.date_range()

```python
# day
print(pd.date_range('2015-07-03', '2015-07-10'))
```

```python
# periods
print(pd.date_range('2015-07-03', periods=8))
```

```python
# time unit
print(pd.date_range('2015-07-03', periods=8, freq='h'))
```

## 23.3 Frequencies and Offsets

| 코드 | 설명 | 코드 | 설명 |
|------|------|------|------|
| D    | 일(day) | B    | 영업일 |
| W    | 주(weekly) | M    | 월말 |
| Q    | 분기말 | A    | 년말 |
| H    | 시간 | T    | 분 |
| S    | 초 | L    | 밀리초 |

```python
# 월간
print(pd.period_range('2015-07', periods=8, freq='M'))

# 2시간 30분 간격
print(pd.timedelta_range(0, periods=6, freq="2h30min"))
```
