# Chapter 22: Vectorized String Operations

## 22.1 Introducing Pandas String Operations

### 일반 Python 방식의 문제점

```python
import numpy as np
import pandas as pd

data = ['peter', 'Paul', 'MARY', 'gUIDO']

# 일반 Python list: roof 필요
result = [s.capitalize() for s in data]
print(result)
```

### 결측치(None)가 있으면 더 복잡해짐

```python
data = ['peter', 'Paul', None, 'MARY', 'gUIDO']

# None 처리필요
result = [s.capitalize() if s is not None else None for s in data]
print(result)
```

### Pandas 방식: **.str** 접근자

```python
names = pd.Series(data)
print(names.str.capitalize())
```

## 22.2 Tables of Pandas String Methods

### 22.2.1 Python 문자열 method and 류사한 method

```python
monte = pd.Series(['Graham Chapman', 'John Cleese', 'Terry Gilliam',
                   'Eric Idle', 'Terry Jones', 'Michael Palin'])

# 소문자 변환
print(monte.str.lower())
```

```python
# 길이
print(monte.str.len())
```

```python
# 시작문자확인
print(monte.str.startswith('T'))
```

### 22.2.2 split()으로 나누기

```python
print(monte.str.split())
```

### 22.3.3 split 결과에서 특정 부분 가져오기

```python
print(monte.str.split().str[-1])
```

### 22.3.4 정규표현식(Regular Expression) 사용

```python
print(monte.str.extract('([A-Za-z]+)', expand=False))
```

```python
# 자음으로 시작하고 자음으로 끝나는 이름 찾기
print(monte.str.findall(r'^[^AEIOU].*[^aeiou]$'))
```

### **22.3.5 get_dummies() - 원-핫 인코딩**

```python
full_monte = pd.DataFrame({'name': monte,
                           'info': ['B|C|D', 'B|D', 'A|C', 'B|D', 'B|C', 'B|C|D']})
full_monte
```

```python
# 구분자 '|'로 분리하여 더미 변수 생성
print(full_monte['info'].str.get_dummies('|'))
```
