# Chapter 13: Introducing Pandas Objects

```python
import numpy as np
import pandas as pd
```

## 13.1 The Pandas Series Object

A Pandas Series is a one-dimensional array of indexed data. It can be created from a
list or array as follows:

```python
data = pd.Series([0.25, 0.5, 0.75, 1.0])
data
```

Series attribute

```python
# 값 확인 (NumPy 배열)
print(data.values)

# 인덱스 확인
print(data.index)
```

💡 data.values ​​is a NumPy array, and data.index is a Pandas Index object.

## 13.2 Series as Generalized NumPy Array

Explicit Index Specification

```python
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=['a', 'b', 'c', 'd'])
data
```

```python
# 인덱스로 접근
print(data['b'])
print(data[['a', 'c']])
```

Non-sequential index

```python
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=[2, 5, 3, 7])
data
```

```python
print(data[5])
```

## 13.3 Series as Specialized Dictionary

Convert Dictionary to Series

```python
population_dict = {'California': 39538223,
                   'Texas': 29145505,
                   'Florida': 21538187,
                   'New York': 20201249,
                   'Pennsylvania': 13002700}
population = pd.Series(population_dict)
population 
```

Here, you can perform typical dictionary-style item access:

```python
population['California']
```

However, unlike dictionaries, Series also supports array-style operations such as slicing:

```python
population['California':'Florida']
```

## 13.4 Constructing Series Objects

Method 1: Create from a list

```python
pd.Series([2, 4, 6])
```

Method 2: Create from a scalar value

```python
pd.Series(5, index=[100, 200, 300])
```

Method 3: Create from a dictionary

```python
pd.Series({2:'a', 1:'b', 3:'c'})
```

Method 4: Explicitly specify an index

```python
pd.Series({2:'a', 1:'b', 3:'c'}, index=[1, 2])
```

## 13.5 The Pandas DataFrame Object

A DataFrame is a two-dimensional labeled data structure.

```python
area_dict = {
    'California': 423967,
    'Texas': 695662,
    'Florida': 170312,
    'New York': 141297,
    'Pennsylvania': 119280
}

area = pd.Series(area_dict)
```

```python
# DataFrame Creation
states = pd.DataFrame({'population': population,
                       'area': area})

print(states)
```

DataFrame attribute

```python
# Row index
print(states.index)

# column index
print(states.columns)
```

## 13.6 DataFrame as Generalized NumPy Array

```python
# accessing value as Numpy array
print(states.values)
```

## 13.7 DataFrame as Specialized Dictionary

```python
# accessing column as key of dictionary
print(states['area'])   # return Series
```

## 13.8 Constructing DataFrame Objects

From single Series

```python
pd.DataFrame(population, columns=['population'])
```

From dictionary list

```python
data = [{'a': i, 'b': 2 * i} for i in range(3)]
print(pd.DataFrame(data))
```

If there are missing keys

```python
pd.DataFrame([{'a': 1, 'b': 2}, {'b': 3, 'c': 4}])
```

From 2D Numpy array

```python
pd.DataFrame(np.random.rand(3, 2),
             columns=['foo', 'bar'],
             index=['a', 'b', 'c'])
```

From a structured array

```python
A = np.zeros(3, dtype=[('A', 'i8'), ('B', 'f8')])
print(A)
pd.DataFrame(A)
```

## 13.9 The Pandas Index Object

Index as Immutable Array

```python
ind = pd.Index([2, 3, 5, 7, 11])
print(ind[1])
print(ind[::2])
```

💡 Index는 불변(immutable)입니다. 한 번 생성하면 수정할 수 없습니다.

```python
ind[1] = 0
# TypeError: Index does not support mutable operations
```

Index as Ordered Set

```python
indA = pd.Index([1, 3, 5, 7, 9])
indB = pd.Index([2, 3, 5, 7, 11])

# 교집합
print(indA.intersection(indB))

# 합집합
print(indA.union(indB))

# 대칭차집합
print(indA.symmetric_difference(indB))
```

💡 Index는 집합 연산을 지원합니다!
