# Chapter 14: Data Indexing and Selection

## 14.1 Data Selection in Series

### 14.1.1 Series as Dictionary

Series give mapping key-value as dictionary

```python
import pandas as pd
import numpy as np

data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=['a', 'b', 'c', 'd'])
print(data)
```

```python
# access by key as dictionary
print(data['b'])

# Check for key existence
print('a' in data)

# Check key list
print(data.keys())

# Check (key, value) pairs with items()
print(list(data.items()))
```

### 14.1.2 Series as One-Dimensional Array

```python
# Slicing
print(data['a':'c'])
```

💡 Note: When slicing with an explicit index, the end index is included!

```python
# slicing
print(data[0:2])
```

💡 When slicing by implicit integer index, the end index is excluded! (NumPy style)

```python
# masking (Boolean array)
print(data[(data > 0.3) & (data < 0.8)])
```

```python
# Fancy indexing
print(data[['a', 'b']])
```

## 14.2 Indexers: loc and iloc

### 14.2.1 loc (Location)

```python
data = pd.Series(['a', 'b', 'c'], index=[1, 3, 5])
print(data)
```

```python
# loc: 항상 명시적 인덱스를 참조
print(data.loc[1])
print(data.loc[1:3])
```

💡 loc includes the end when slicing based on index labels.

### 14.2.2 iloc (Integer Location)

```python
# iloc: 항상 암시적 Python 스타일 인덱스를 참조
print(data.iloc[1])
print(data.iloc[1:3])
```

💡 iloc is based on the index position. Same as NumPy style!

## 14.3 Data Selection in DataFrames

### 14.3.1 DataFrame as Dictionary

```python
area = pd.Series({'California': 423967, 'Texas': 695662,
                  'Florida': 170312, 'New York': 141297,
                  'Pennsylvania': 119280})
pop = pd.Series({'California': 39538223, 'Texas': 29145505,
                 'Florida': 21538187, 'New York': 20201249,
                 'Pennsylvania': 13002700})
data = pd.DataFrame({'area': area, 'pop': pop})
data
```

```python
print(data['area'])
```

💡 Accessing properties is convenient, but cannot be used if the column name conflicts with the method name or contains spaces!

```python
# add new column (like dictionary)
data['density'] = data['pop'] / data['area']
data
```

### 14.3.2 DataFrame as Two-Dimensional Array

```python
print(data.values)
```

```python
print(data.T)
```

```python
# By iloc indexing as Numpy style
print(data.iloc[:3, :2])
```

```python
# By loc indexing as label style
print(data.loc['Florida':'New York', 'area':'pop'])
```

```python
# By loc masking and fancy indexing
print(data.loc[data.density > 120, ['pop', 'density']])
```

```python
# Modify value
data.iloc[0, 2] = 90
data
```

## 14.4 Additional Indexing Conventions

### 14.4.1 Slicing References Rows

```python
print(data['Florida':'New York'])
```

### 14.4.2 Masking References Rows

```python
print(data[data.density > 120])
```

💡 DataFrame에서 직접 indexing할 때는 렬이 아니라 행이 선택된다는 점을 기억!
