### Chapter 7: Aggregations: min, max, and Everything in Between

**7.1 Summing the Values in an Array**

```python
import numpy as np

# Python sum() vs Numpy sum()
L = np.random.random(100)

print(sum(L))
print(np.sum(L))

big_array = np.random.random(1_000_000)

%timeit sum(big_array)
%timeit np.sum(big_array)
```

**7.2 Minimum and Maximum**

```python
# Python function
min(big_array), max(big_array)

# Numpy function
np.min(big_array), np.max(big_array)

# array method
big_array.min()
big_array.max()
big_array.sum()
```

**7.3 Multidimensional Aggregates**

```python
# Creation 2D array
rng = np.random.default_rng(seed=1701)
M = rng.integers(0, 10, (3, 4))
print(M)

# Sum all
print(M.sum())

# axis=0: 렬방향집계 (각 렬의 합)
print(M.sum(axis=0))

# axis=1: 행방향집계 (각 행의 합)
print(M.sum(axis=1))

# 각 렬의 최소값
print(M.min(axis=0))

# 각 행의 최대값
print(M.max(axis=1))
```

**7.5 Example: What Is the Average Height of US Presidents?**

```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/jakevdp/PythonDataScienceHandbook/master/notebooks/data/president_heights.csv"
data = pd.read_csv(url)
heights = np.array(data['height(cm)'])
print(heights)

print("Mean height:         ", heights.mean())
print("Standard deviation:  ", heights.std())
print("Minimum height:      ", heights.min())
print("Maximum height:      ", heights.max())

# 백분위수
print("25th percentile:     ", np.percentile(heights, 25))
print("Median:              ", np.median(heights))
print("75th percentile:     ", np.percentile(heights, 75))
```

*visualization*

```python
import matplotlib.pyplot as plt

plt.hist(heights)
plt.title('Height Distribution of US Presidents')
plt.xlabel('height (cm)')
plt.ylabel('number');
```
