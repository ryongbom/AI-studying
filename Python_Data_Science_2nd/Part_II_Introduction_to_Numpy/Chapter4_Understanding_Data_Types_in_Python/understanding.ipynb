# Chapter 4. Understanding Data Types In Python

Python is a dynamically typed language.

```python
# C/C++과의 차이

# C++에서는 이렇게 타입을 선언해야 함
# int x = 10;
# x = "hello";  # Error!

# Python에서는 타입 선언 없이 자유롭게 변경 가능
x = 10      # 정수
x = "hello" # 문자열 - 문제없음!
```

Python integers are not simple integers.

```python
# Python 정수의 내부 구조
x = 10000

# Python 정수는 C 구조체로 구현되어 있음
# - ob_refcnt: 참조 카운트
# - ob_type: 타입 정보
# - ob_size: 크기
# - ob_digit: 실제 값

# 이 때문에 C에 비해 메모리 사용량이 큼
import sys
print(sys.getsizeof(10000))  # 28바이트 (C의 int는 4바이트)
```
