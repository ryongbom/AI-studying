## 6.1 The Slowness of Loops

Python's default implementation (CPython) processes some operations very slowly. This is partly due to the dynamic and interpreted nature of the language. Because the types are flexible, sequences of operations cannot be compiled into efficient machine code like in languages ​​such as C or Fortran.

```python
import numpy as np
rng = np.random.default_rng(seed=1701)

def compute_reciprocals(values):
    output = np.empty(len(values))
    for i in range(len(values)):
        output[i] = 1.0 / values[i]
    return output

values = rng.integers(1, 10, size=5)
print(values)
print(compute_reciprocals(values))
```

This implementation feels natural to those with a C or Java background. However, if you measure the execution time for large inputs, you will find that it is very slow:

```python
big_array = rng.integers(1, 100, size=1_000_000)
%timeit compute_reciprocals(big_array)
```

## 6.2 Introducing Ufuncs (Ufuncs 소개)

NumPy provides a convenient interface for these statically typed compiled routines. This is called a vectorized operation.

```python
# vectorized operation
%timeit (1.0 / big_array)
```

Vectorized NumPy operations are implemented through ufunc. ufunc supports operations between arrays as well as between scalars and arrays.

```python
# 스칼라와 배렬연산
np.array(5) / np.arange(1, 6)

# 다차원 배렬연산
x = np.arange(9).reshape((3, 3))
2 ** x
```

*Key point: Whenever you encounter a Python loop, consider whether it can be replaced with a vectorized representation!*

## 6.3 Exploring NumPy's Ufuncs

Ufuncs는 단항(unary) 과 이항(binary) 두 가지 형태로 존재합니다.

* Array Arithmetic

```python
x = np.arange(4)
print("x     =", x)
print("x + 5 =", x + 5)
print("x - 5 =", x - 5)
print("x * 2 =", x * 2)
print("x / 2 =", x / 2)
print("x // 2 =", x // 2)

print("-x     =", -x)
print("x ** 2 =", x ** 2)
print("x % 2  =", x % 2)
```

* Absolute Value

```python
x = np.array([-2, -1, 0, 1, 2])
abs(x)              # array([2, 1, 0, 1, 2])
np.absolute(x)      # array([2, 1, 0, 1, 2])
np.abs(x)           # array([2, 1, 0, 1, 2])

# 복소수크기
x = np.array([3 - 4j, 4 - 3j, 2 + 0j, 0 + 1j])
abs(x)
```

* Trigonometric Functions (삼각함수)

```python
theta = np.linspace(0, np.pi, 3)
print("theta      =", theta)
print("sin(theta) =", np.sin(theta))
print("cos(theta) =", np.cos(theta))
print("tan(theta) =", np.tan(theta))

# 거꿀삼각함수
x = [-1, 0, 1]
print("arcsin(x) =", np.arcsin(x))
print("arccos(x) =", np.arccos(x))
print("arctan(x) =", np.arctan(x))
```
